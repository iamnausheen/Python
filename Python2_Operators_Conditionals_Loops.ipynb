```python
# Arithmetic - mathematical operations
print(5+6)
print(5-6)
print(5/6)
print(5*6)
print(5/2)
print(5//2)   # INTEGER DIVISION - converts float answer to integer
print(5%2)    # Remainder
print(5**2)
```

```python
# Relational Operators - for comparison
print(4>5)
print(4<5)
print(5>=4)
print(4<=4)
print(4==4)
print(4!=4)
```

```python
# Logical operators - for logic building

# and
print(1 and 0)
print(0 and 1)
print(0 and 0)
print(1 and 1)

# or
print(1 or 0)
print(0 or 1)
print(1 or 1)
print(0 or 0)

# not
print(not 0)
print(not 1)
```

```python
# Bitwise operators - work on bits

# bitwise and
print(2 & 3)
# 2 = 010
# 3 = 011
# 2 & 3 = (0 & 0) (1 & 1) (0 & 1) = 010 = 2

# bitwise or
print(2 | 3)
# 2 or 3 = 011 = 3

# bitwise xor
print(1 ^ 1)
print(1 ^ 0)
print(0 ^ 1)
print(1 ^ 1)

print(2 ^ 3)
# 2 ^ 3 = 1

# bitwise not
print(~0)
print(~1)
print(~2)
print(~3)

# Bitwise left shift
print(4 >> 2)
# 4 = 000.......100
# 4 >> 2 = 0000.....1000 = 8

# Bitwise right shift
print(4 << 2)
# 4 = 000.......100
# 4 >> 2 = 0000.....010 = 2
```

```python
# Assignment operators
a = 2
a = a + 2
a += 2
print(a)
a *= 2
print(a)
a /= 2
print(a)
a -= 2
print(a)

# C - a++, a--, ++a, --a
```

```python
# Membership Operators
# in / not in

print('D' in 'Delhi')   # Faster than loop
print(1 in [1, 2, 3, 4, 5])

print('D' not in 'Delhi')   # Faster than loop
print(1 not in [1, 2, 3, 4, 5])

print('J' in 'Delhi')   # Faster than loop
print(0 in [1, 2, 3, 4, 5])

print('J' not in 'Delhi')   # Faster than loop
print(0 not in [1, 2, 3, 4, 5])

```

```python
# Find the sum of a 3 digit number by the user

num = int(input("3 digit number: "))

ones_place = num%10
num = num//10

tenths_place = num%10
num = num//10

hundreths_place = num

total = ones_place + tenths_place + hundreths_place
print(total)


# using loop
num = int(input("3 digit number: "))
total = 0
while num:
  d = num%10
  total += d
  num = num//10
print(total)

# using membership operator and loop
num = input("3 digit number: ")
total = 0
for digit in num:
  total += int(digit)

print(total)
```

```python
# Find the minimum of 3 numbers

num1 = int(input("First number: "))
num2 = int(input("Second number: "))
num3 = int(input("Third number: "))

minm = None
if num1 <= num2 and num1 <= num3:
  minm = num1
elif num2 <= num3:
  minm = num2
else:
  minm = num3
print("smallest number = ", minm)

# finding smallest number
if num1 <= num2:
  if num1 <= num3:
    minm = num1
  else:
    minm = num3
else:
  if num2 <= num3:
    minm = num2
  else:
    minm = num3
print("smallest number = ", minm)
```

```python
# creating menus using if-else
# CLI applications

# menu driven calc, operations -> +, -, *, /

num1 = int(input("First number: "))
num2 = int(input("Second number: "))

operation = input("Operation:")

if operation == '+':
  print(num1 + num2)
elif operation == '-':
  print(num1 - num2)
elif operation == '*':
  print(num1*num2)
elif operation == '/':
  print(num1/num2)
else:
  print("Invalid Operation")
```

```python
# Guessing game

# generate a random integer between 1 and 100
# ask user to guess until actual number guessed
# notify if greater or less than actual number

import random
num = random.randint(1, 100)
score = 1

guess = int(input("Guess the number: "))

while guess != num:
  if guess > num:
    guess = int(input("Guess Lower: "))
  else:
    guess = int(input("Guess Higher: "))
  score += 1
else:
  print("You GOT IT ! Score = ", score)

```
